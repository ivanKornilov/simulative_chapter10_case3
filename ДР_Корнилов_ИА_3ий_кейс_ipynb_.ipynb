### **Посчитайте Retention 15 дня (в процентах) для пользователей, зарегистрированных в январе**

Cохраните результат в переменную `retention_15_day`

**Примечание:** результат округлите до 5 знаков после запятой

```python
# Ваше решение
from datetime import datetime, timedelta
import csv

# Загрузка регистраций
reg_by_user = {}
with open('registrations.csv') as f:
   reader = csv.DictReader(f, delimiter=';')
   for row in reader:
       user_id = row['user_id']
       try:
           reg_date = datetime.strptime(row['registration_date'], '%Y-%m-%d').date()
       except ValueError:
           continue
       reg_by_user[user_id] = reg_date

# Загрузка входов и группировка по user_id
entries_by_user = {}
with open('entries.csv', encoding='utf-8') as f:
   reader = csv.DictReader(f, delimiter=';')
   for row in reader:
       user_id = row['user_id']
       try:
           entry_date = datetime.strptime(row['entry_date'], '%Y-%m-%d').date()
       except ValueError:
           continue
       entries_by_user.setdefault(user_id, set()).add(entry_date)

# Пользователи, зарегистрированные в январе (любой год)
jan_user_ids = [uid for uid, d in reg_by_user.items() if d.month == 1]

total_jan_users = len(jan_user_ids)
retained_users = 0

for user_id in jan_user_ids:
   reg_date = reg_by_user[user_id]
   target_date = reg_date + timedelta(days=15)
   if target_date in entries_by_user.get(user_id, set()):
       retained_users += 1

retention_15_day = (retained_users / total_jan_users * 100) if total_jan_users > 0 else 0.0
print(f"retention_15_day = {retention_15_day:.5f}%")
retention_15_day = round(retention_15_day, 5)
```

```python
#@title ✏️ Проверка: чтобы проверить свое решение запустите код в этой ячейке
# Открываем файл с правильными ответами
with open('metrics.txt', 'r') as f:
    answers = f.read().split('\n')

correct_answer = float(answers[0])

try:
    assert retention_15_day == correct_answer
except AssertionError:
    print('Ответы не совпадают')
else:
    print('Поздравляем, Вы справились!')
```

### **Посчитайте Lifetime по всем пользователям, посчитанный как интеграл от n-day retention**

Сохраните результат в переменную `lifetime`

**Примечание:** результат округлите до 5 знаков после запятой

```python
import csv
from datetime import datetime, timedelta
from collections import defaultdict

reg_by_user = {}
with open('registrations.csv', encoding='utf-8') as f:
   reader = csv.DictReader(f, delimiter=';')
   for row in reader:
       user_id = row['user_id']
       try:
           reg_date = datetime.strptime(row['registration_date'], '%Y-%m-%d').date()
       except ValueError:
           continue
       reg_by_user[user_id] = reg_date

entries_by_user = defaultdict(set)
with open('entries.csv', encoding='utf-8') as f:
   reader = csv.DictReader(f, delimiter=';')
   for row in reader:
       user_id = row['user_id']
       try:
           entry_date = datetime.strptime(row['entry_date'], '%Y-%m-%d').date()
       except ValueError:
           continue
       entries_by_user[user_id].add(entry_date)

total_retention_sum = 0.0
num_users = len(reg_by_user)

for user_id, reg_date in reg_by_user.items():
   entries = entries_by_user.get(user_id, set())
   last_entry = max(entries) if entries else reg_date


   max_days = min((last_entry - reg_date).days, 180)


   for n in range(0, max_days + 1):
       target = reg_date + timedelta(days=n)
       total_retention_sum += 1.0 if target in entries else 0.0


lifetime = total_retention_sum / num_users if num_users > 0 else 0.0
lifetime = round(lifetime, 5)
print(f"lifetime = {lifetime}")

```

```python
#@title ✏️ Проверка: чтобы проверить свое решение запустите код в этой ячейке
correct_answer = float(answers[2])

try:
    assert lifetime == correct_answer
except AssertionError:
    print('Ответы не совпадают')
else:
    print('Поздравляем, Вы справились!')
```

### **Посчитайте Churn rate 29 дня (в долях), посчитанный по всем пользователям**

Сохраните результат в переменную `churn_29`.  
Если будете считать CR от Retention, ведите расчет от Rolling Retention, таким образом, мы получим всех, кто не заходил в 29 день и после. Ведя расчет от обычного Retention, наоборот - получим только CR в 29 день.


```python
import csv
from datetime import datetime
from collections import defaultdict

reg_by_user = {}
with open('registrations.csv', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f, delimiter=';')
    for row in reader:
        user_id = row['user_id']
        try:
            reg_date = datetime.strptime(row['registration_date'], '%Y-%m-%d').date()
        except ValueError:
            continue
        reg_by_user[user_id] = reg_date

entries_by_user = defaultdict(set)
with open('entries.csv', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f, delimiter=';')
    for row in reader:
        user_id = row['user_id']
        try:
            entry_date = datetime.strptime(row['entry_date'], '%Y-%m-%d').date()
        except ValueError:
            continue
        entries_by_user[user_id].add(entry_date)

num_users = len(reg_by_user)
churn_count = 0

for user_id, reg_date in reg_by_user.items():
    entries = entries_by_user.get(user_id, set())
    has_entry_29_or_later = any((e - reg_date).days >= 29 for e in entries)
    if not has_entry_29_or_later:
        churn_count += 1

churn_29 = churn_count / num_users if num_users > 0 else 0.0

print(f"churn_29 = {churn_29}")
```

### **Посчитайте Mau, Wau, Dau за последний месяц/неделю/день записей**

### **Посчитайте Mau, Wau, Dau усредненные**

Сохраните результат в переменные `avg_mau`, `avg_wau`, `avg_dau` соответственно

**Примечание:** результаты округлите до 5 знаков после запятой

```python
import csv
from datetime import datetime
from collections import defaultdict

entries = []
with open('entries.csv', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f, delimiter=';')
    for row in reader:
        user_id = row['user_id']
        try:
            entry_date = datetime.strptime(row['entry_date'], '%Y-%m-%d').date()
        except ValueError:
            continue
        entries.append((user_id, entry_date))

month_to_users = defaultdict(set)  # (year, month) -> set(user_id)
week_to_users = defaultdict(set)   # (year, week) -> set(user_id)
day_to_users = defaultdict(set)   # date -> set(user_id)

for uid, d in entries:
    month_to_users[(d.year, d.month)].add(uid)
    iso = d.isocalendar()
    week_to_users[(iso.year, iso.week)].add(uid)
    day_to_users[d].add(uid)

avg_mau = round(
    sum(len(s) for s in month_to_users.values()) / len(month_to_users) if month_to_users else 0.0,
    5
)

avg_wau = round(
    sum(len(s) for s in week_to_users.values()) / len(week_to_users) if week_to_users else 0.0,
    5
)

avg_dau = round(
    sum(len(s) for s in day_to_users.values()) / len(day_to_users) if day_to_users else 0.0,
    5
)

print(f"avg_mau = {avg_mau}")
```
